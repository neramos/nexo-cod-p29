```python
import numpy as np

def resonance_measure(r, N=1000, modulus=7):
    """Computes resonance measure R(r) for growth ratio r."""
    steps = np.arange(1, N+1)
    distances = np.abs(steps * r - np.round(steps * r / modulus) * modulus)
    return np.min(distances)

# Example usage with your weights
ratios = {
    'Golden (φ)': (1 + np.sqrt(5))/2,
    'Silver': 1 + np.sqrt(2),
    'Bronze': 1 + np.sqrt(3),
    'Copper': 1 + np.sqrt(5),
    'Euler': np.e,
    'Pi': np.pi
}

print("Resonance Measure (lower = stronger resonance):")
for name, r in ratios.items():
    R = resonance_measure(r)
    print(f"{name:<20} | R = {R:.4f}")
```
