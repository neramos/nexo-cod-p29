```python
"""
octonion_fano_plane.py
======================
Visualizing the Octonion Multiplication using the Fano Plane.
Demonstrates that Octonions (8D) are a FINITE combinatorial structure,
not an infinite continuous expansion.

Author: Néstor E. Ramos
For: The "Algebraic Horizon" chapter.
"""
import matplotlib.pyplot as plt
import numpy as np

print("=" * 70)
print("OCTONION FINITISM: THE FANO PLANE LIMIT")
print("=" * 70)

# The 7 imaginary units e1..e7
units = [1, 2, 3, 4, 5, 6, 7]

# Multiplication rules (Cyclic Fano Plane)
# (i, j, k) means e_i * e_j = e_k
triples = [
    (1, 2, 3), (1, 4, 5), (1, 6, 7),
    (2, 4, 6), (2, 7, 5),
    (3, 4, 7), (3, 6, 5)
]

print("Octonion Structure (Base: {1, i, j, k, l, il, jl, kl})")
print("Total Dimensions: 8 (1 Real + 7 Imaginary)")
print("This is a FINITE set. No 'infinite' dimensions.")
print("Beyond this (Sedenions, 16D), the algebra breaks (Zero Divisors).")

# Plot Fano Plane
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_aspect('equal')
ax.axis('off')

# Fano Plane Geometry
points = {
    1: (0, 1), 2: (0.86, 0.5), 3: (0.86, -0.5),
    4: (0, -1), 5: (-0.86, -0.5), 6: (-0.86, 0.5), 7: (0, 0)
}

colors = plt.cm.viridis(np.linspace(0, 1, 8))

# Draw Lines (Triples)
for t in triples:
    i, j, k = t
    # Draw circle or line connecting i, j, k
    # Fano plane: 1-2-3 is a line (circle at infinity usually, here simplified)
    # For simplicity, we draw the triangle 1-2-3
    if t == (1, 2, 3):
        circle = plt.Circle((0, 0), 1, fill=False, color='gold', linewidth=3, linestyle='--', label='e1-e2-e3 (Infinity)')
        ax.add_patch(circle)
    else:
        xs = [points[i][0], points[j][0], points[k][0], points[i][0]]
        ys = [points[i][1], points[j][1], points[k][1], points[i][1]]
        ax.plot(xs, ys, color='orange', linewidth=2, alpha=0.7)

# Draw Points
for i, pos in points.items():
    ax.plot(pos[0], pos[1], 'o', markersize=20, color=colors[i], label=f'e{i}')
    ax.text(pos[0], pos[1] + 0.15, str(i), fontsize=14, ha='center', fontweight='bold')

plt.title("The Fano Plane: Finite Structure of Octonions (8D)\nBeyond 8D (Sedenions), Algebra Breaks (Zero Divisors)",
          fontsize=14, fontweight='bold', pad=20)
plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1))
plt.tight_layout()
plt.savefig('octonion_fano_plane.png', dpi=300, bbox_inches='tight')
print("\n✓ Saved: octonion_fano_plane.png")
plt.show()
```
