```python
"""
radix_invariance_proof.py
==========================
Proves that the Planck Voxel's interaction phase space (144 channels)
and information entropy are invariant across different numerical bases.
"""

import numpy as np
import matplotlib.pyplot as plt
import math

print("=" * 70)
print("RADIX INVARIANCE PROOF: Base-Independent Planck Substrate")
print("=" * 70)

# The fundamental physical constants (Base-independent)
K_max = 24          # 4D spatial topology
N_dof = 6           # Internal registers (1+2+3 partition)
Fano_points = 7     # Total Fano states
Usable_Fano = 6     # Active quantum states
Total_channels = K_max * Usable_Fano  # 144

print(f"\n[Fundamental Physical Invariants]")
print(f"  Spatial Topology (K_max): {K_max}")
print(f"  Internal Registers (N_dof): {N_dof}")
print(f"  Fano Geometry: {Usable_Fano} usable / {Fano_points} total")
print(f"  Total Interaction Channels: {Total_channels}")

# =============================================================================
# BASE CONVERSION & ENTROPY ANALYSIS
# =============================================================================

def analyze_base(base, name):
    """
    Analyze how the 144 channels are represented in a given base.
    """
    # How many "digits" in this base are needed to represent 144 states?
    # We need base^digits >= 144
    digits_needed = math.ceil(math.log(Total_channels, base))

    # Total possible states in this representation
    total_representable = base ** digits_needed

    # Information Entropy (Shannon) in bits. This MUST be invariant.
    entropy_bits = math.log2(Total_channels)

    # The probability of a specific U(1) interaction
    alpha_prob = 1 / Total_channels

    print(f"\n--- Analysis: {name} (Base-{base}) ---")
    print(f"  Digits required to encode 144 channels: {digits_needed}")
    print(f"  Total representable states in {digits_needed} digits: {total_representable}")
    print(f"  Wasted states (overflow): {total_representable - Total_channels}")
    print(f"  Information Entropy: {entropy_bits:.4f} bits (INVARIANT)")
    print(f"  Interaction Probability (alpha): {alpha_prob:.8f} (INVARIANT)")

    return digits_needed, entropy_bits

# Test different bases
bases = [
    (2, "Binary"),
    (8, "Octal"),
    (10, "Decimal (Base-10)"),
    (12, "Duodecimal"),
    (16, "Hexadecimal"),
    (60, "Sexagesimal (Babylonian)")
]

entropies = []
base_names = []

for base, name in bases:
    digits, entropy = analyze_base(base, name)
    entropies.append(entropy)
    base_names.append(f"{name}\n(Base-{base})")

# =============================================================================
# VISUALIZATION
# =============================================================================

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle("Radix Invariance of the Planck Substrate", fontsize=14, fontweight='bold')

# Left: Digits required per base
ax1.bar(base_names, [math.ceil(math.log(Total_channels, b)) for b, _ in bases],
        color='steelblue', edgecolor='black')
ax1.set_ylabel('Number of Digits Required', fontsize=11)
ax1.set_title('Encoding Efficiency Across Bases', fontweight='bold')
ax1.grid(axis='y', alpha=0.3)
ax1.text(0.5, 0.9, "The number of digits changes,\nbut the information content\nremains exactly 144 states.",
         transform=ax1.transAxes, ha='center', fontsize=10,
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

# Right: Shannon Entropy (The Invariant)
ax2.bar(base_names, entropies, color='seagreen', edgecolor='black')
ax2.axhline(y=entropies[0], color='red', linestyle='--', linewidth=2,
            label=f'Invariant Entropy: {entropies[0]:.4f} bits')
ax2.set_ylabel('Shannon Entropy (bits)', fontsize=11)
ax2.set_title('Information Content is Base-Independent', fontweight='bold')
ax2.legend()
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
fig.savefig("radix_invariance_proof.png", dpi=300, bbox_inches='tight')
print("\n✓ Saved: radix_invariance_proof.png")
plt.show()

print("\n" + "=" * 70)
print("CONCLUSION")
print("=" * 70)
print("The simulation proves that while the 'packaging' of the data changes")
print("(e.g., 8 bits in binary vs 2 digits in hex), the actual physical")
print("information content (7.17 bits) and the interaction probability (1/144)")
print("are absolute invariants of the Fano/4D topology.")
print("=" * 70)
```
