```python
"""
finitist_black_hole_smooth.py
==============================
Visualizing the 'Horizon Smear': Structural Completion without Singularity.
Replaces the shifted pole with a smooth saturation function.

Author: Néstor E. Ramos
For: "The Pixelated Universe" - Black Hole Finitism
"""

import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print("=" * 70)
print("FINITIST BLACK HOLE: THE HORIZON SMEAR (NO SINGULARITY)")
print("=" * 70)

# =============================================================================
# CONSTANTS
# =============================================================================
d_max = 9
epsilon = 1.0 / d_max  # Finite resolution limit
x = np.linspace(0.5, 1.5, 2000)  # x = R_g / r

# =============================================================================
# FUNCTIONS
# =============================================================================

# Continuum GR: Infinite pole at x=1
g_rr_cont = 1.0 / (1.0 - x)

# Finitist Substrate: Smooth Saturation (Regularization)
# The denominator never goes to zero; it bottoms out at epsilon
g_rr_fin = 1.0 / np.sqrt((1.0 - x)**2 + epsilon**2)

# =============================================================================
# VISUALIZATION
# =============================================================================

def plot_smooth_saturation():
    fig, (ax_main, ax_inset) = plt.subplots(1, 2, figsize=(18, 7))

    # --- Main Plot ---
    ax_main.plot(x, g_rr_cont, 'r-', linewidth=1.5, label='Continuum (Singularity)', alpha=0.5)
    ax_main.plot(x, g_rr_fin, 'g-', linewidth=2.5, label=f'Finitist Substrate (Peak = {d_max})')
    ax_main.axhline(y=d_max, color='green', linestyle=':', linewidth=2, label='Max Causal Density')

    # Highlight the smeared region
    ax_main.axvspan(1.0 - epsilon, 1.0 + epsilon, color='yellow', alpha=0.2, label='Horizon Smear')

    ax_main.set_title("Schwarzschild Metric: Singularity vs. Smooth Saturation", fontsize=14, fontweight='bold')
    ax_main.set_xlabel("Proportion $x = R_g / r$", fontsize=12)
    ax_main.set_ylabel("$g_{rr}$ (Causal Density)", fontsize=12)
    ax_main.set_ylim(0, d_max + 2)
    ax_main.legend(fontsize=11)
    ax_main.grid(True, alpha=0.3)

    # --- Inset: The Peak ---
    ax_inset.plot(x, g_rr_fin, 'g-', linewidth=2)
    ax_inset.fill_between(x, 0, g_rr_fin, color='green', alpha=0.1)
    ax_inset.axhline(y=d_max, color='red', linestyle='--', linewidth=1.5, label='Saturation Cap')

    ax_inset.set_xlim(0.95, 1.05)
    ax_inset.set_ylim(d_max - 1, d_max + 0.5)
    ax_inset.set_title("The Horizon Smear\nFinite Thickness Δx ≈ 1/d_max", fontsize=12, fontweight='bold')
    ax_inset.legend(fontsize=10)
    ax_inset.grid(True, alpha=0.3)

    # Clean text (Unicode safe, no LaTeX parsing issues)
    ax_inset.text(0.96, d_max - 0.5, "No Vertical Line.\nNo Infinity.\nJust Max Density.",
                 fontsize=10, color='darkgreen', fontweight='bold')

    plt.suptitle("Computational Finitism: The Black Hole is a Saturated Cluster", fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

    print("\n" + "=" * 70)
    print("RESULT: STRUCTURAL COMPLETION")
    print("=" * 70)
    print(f"• Max Density: {d_max} (at x=1)")
    print(f"• Horizon Thickness: Δx ≈ {epsilon:.3f}")
    print("• The 'Singularity' is resolved into a smooth peak.")
    print("• A Black Hole is a region of maximum causal density.")
    print("=" * 70)

if __name__ == "__main__":
    plot_smooth_saturation()
```
