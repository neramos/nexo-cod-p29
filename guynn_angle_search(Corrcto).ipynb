```python
"""
guynn_angle_search.py
======================
Search for a formula for Guynn's angle θ_a1 = 37.46731118°.

Inputs (known framework constants):
  phi = (1+sqrt(5))/2 = golden ratio
  alpha = 1/137
  a_e = 0.00115965 (anomalous g-factor)
  pi_disc = 22/7 (discrete pi)

We look for expressions of the form:
  sin(θ_a1) = f(phi, alpha, pi_disc, a_e, ...)
"""

import numpy as np
from itertools import product

# Observed value
theta_a1_obs = 37.46731118
sin_a1_obs = np.sin(np.radians(theta_a1_obs))
cos_a1_obs = np.cos(np.radians(theta_a1_obs))
tan_a1_obs = np.tan(np.radians(theta_a1_obs))

print(f"Observed θ_a1 = {theta_a1_obs}°")
print(f"  sin = {sin_a1_obs:.7f}")
print(f"  cos = {cos_a1_obs:.7f}")
print(f"  tan = {tan_a1_obs:.7f}")
print()

# Framework constants
phi = (1 + np.sqrt(5)) / 2
alpha = 1 / 137
a_e = 0.00115965
pi_disc = 22 / 7

print(f"Framework constants:")
print(f"  φ       = {phi:.7f}")
print(f"  1/φ     = {1/phi:.7f}")
print(f"  φ²      = {phi**2:.7f}")
print(f"  1/φ²    = {1/phi**2:.7f}")
print(f"  α       = {alpha:.7f}")
print(f"  a_e     = {a_e:.7f}")
print(f"  π_disc  = {pi_disc:.7f}")
print()

# ---------------------------------------------------------
# CANDIDATE BASE ANGLES
# ---------------------------------------------------------
print("Candidate base angles and their sines:")
candidates_base = {
    "45°/2 = 22.5°":  22.5,
    "60° - 22.5°":    37.5,
    "arctan(1/φ)":    np.degrees(np.arctan(1/phi)),
    "arccos(1/φ²)":   np.degrees(np.arccos(1/phi**2)),
    "arccos(1/√5)":   np.degrees(np.arccos(1/np.sqrt(5))),
    "arccos(2/3)":    np.degrees(np.arccos(2/3)),
    "60° - arctan(1/φ)": 60 - np.degrees(np.arctan(1/phi)),
    "60° - arccos(1/φ²)": 60 - np.degrees(np.arccos(1/phi**2)),
}
for name, ang in candidates_base.items():
    print(f"  {name:30s} = {ang:.6f}°, sin = {np.sin(np.radians(ang)):.6f}")
print()

# ---------------------------------------------------------
# SEARCH: expressions of form sin(θ_a1) = A - B*C
# ---------------------------------------------------------
print("=" * 60)
print("SEARCH: expressions of form A - B·C matching sin(θ_a1)")
print("=" * 60)
print()

target = sin_a1_obs
print(f"Target sin(θ_a1) = {target:.7f}")
print()

# Base values to try
bases = {
    "1/φ":   1/phi,
    "1/φ²":  1/phi**2,
    "1/φ³":  1/phi**3,
    "1/φ⁴":  1/phi**4,
    "1/√5":  1/np.sqrt(5),
    "√5-2":  np.sqrt(5) - 2,
    "√2-1":  np.sqrt(2) - 1,
    "α":     alpha,
    "a_e":   a_e,
}

hits = []
for name_A, A in bases.items():
    for name_B, B in bases.items():
        for name_C, C in bases.items():
            val = A - B * C
            if abs(val - target) < 1e-5:
                hits.append((name_A, name_B, name_C, val))

if hits:
    print("Matches found:")
    for h in hits[:20]:
        print(f"  {h[0]} - {h[1]}·{h[2]} = {h[3]:.7f}  (error {abs(h[3]-target):.2e})")
else:
    print("No exact matches found in base-value combinations.")
print()

# ---------------------------------------------------------
# CHECK: what if the angle is exactly 37.5° with a correction?
# ---------------------------------------------------------
print("=" * 60)
print("STRUCTURE: 37.5° + correction")
print("=" * 60)
print()

correction = theta_a1_obs - 37.5
print(f"θ_a1 - 37.5° = {correction:.7f}°")
print()

# Compare correction to combinations of α, a_e, φ
print("Candidate corrections:")
candidates_corr = {
    "α":                alpha,
    "a_e":              a_e,
    "5α/4":             5*alpha/4,
    "5a_e/4":           5*a_e/4,
    "α/φ":              alpha/phi,
    "α/φ²":             alpha/phi**2,
    "a_e/φ":            a_e/phi,
    "a_e·α":            a_e*alpha,
    "α/(2π_disc)":      alpha/(2*pi_disc),
    "1/(4·168)":        1/(4*168),
    "1/(4·168·φ)":      1/(4*168*phi),
    "α²/π":             alpha**2/np.pi,
}
for name, val in candidates_corr.items():
    print(f"  {name:20s} = {val:.7f}  (ratio to observed: {correction/val:.4f})")
print()

# ---------------------------------------------------------
# CHECK: 60° - 37.4673° = 22.5327°. What is 22.5327°?
# ---------------------------------------------------------
print("=" * 60)
print("STRUCTURE: θ_ea = 22.5° + correction")
print("=" * 60)
print()

corr_ea = 22.53268882 - 22.5
print(f"θ_ea - 22.5° = {corr_ea:.7f}°")
print(f"θ_ea / 22.5° = {22.53268882 / 22.5:.7f}")
print()

# What is the ratio?
ratio = 22.53268882 / 22.5
print(f"Ratio: {ratio:.7f}")
print(f"1 - ratio: {1 - ratio:.7f}")
print(f"(ratio - 1) / α = {(ratio-1)/alpha:.4f}")
print(f"(ratio - 1) / a_e = {(ratio-1)/a_e:.4f}")
print(f"(ratio - 1) / (5a_e/4) = {(ratio-1)/(5*a_e/4):.4f}")
print()

print("=" * 60)
print("HONEST CONCLUSION")
print("=" * 60)
print()
print("We have strong numerical structure:")
print(f"  θ_ea = 45°/2 + correction")
print(f"  correction ≈ 0.0327°")
print()
print("But no algebraic derivation of the correction.")
print("The factor 5/4 in the candidate fit has no justification.")
print()
print("Need: Guynn's original integral formula to see what")
print("      produces the exact value.")
```
