```python
"""
time_dilation_finitist.py
==========================
Demonstrating that time dilation is FINITE at v=c in computational finitism.
Replaces continuum divergence with smooth saturation.

Author: Néstor E. Ramos
For: "The Pixelated Universe" - Finite Velocity Dynamics
"""

import numpy as np
import matplotlib.pyplot as plt

print("=" * 70)
print("TIME DILATION IN FINITIST FRAMEWORK: NO INFINITY AT v=c")
print("=" * 70)

# =============================================================================
# PARAMETERS
# =============================================================================
d_max = 9
epsilon = 1.0 / d_max  # Finite resolution limit

# Velocity range (approaching c)
beta = np.linspace(0, 1.0, 1000)  # beta = v/c

# =============================================================================
# TIME DILATION FUNCTIONS
# =============================================================================

# Standard continuum (divergent at beta=1)
def gamma_continuum(beta):
    return 1.0 / np.sqrt(1 - beta**2 + 1e-15)  # Small offset to avoid division by zero

# Finitist regularization (smooth saturation)
def gamma_finitist(beta, epsilon):
    return 1.0 / np.sqrt((1 - beta**2)**2 + epsilon**2)

# Alternative: Linear regularization (analogous to black hole)
def gamma_finitist_linear(beta, epsilon):
    return 1.0 / np.sqrt(np.abs(1 - beta**2) + epsilon)

# =============================================================================
# VISUALIZATION
# =============================================================================

def plot_time_dilation():
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

    # Left: Full range
    ax1.plot(beta, gamma_continuum(beta), 'r-', linewidth=2, label='Continuum (Divergent)', alpha=0.6)
    ax1.plot(beta, gamma_finitist(beta, epsilon), 'g-', linewidth=2.5, label=f'Finitist (ε={epsilon:.3f})')
    ax1.plot(beta, gamma_finitist_linear(beta, epsilon), 'b--', linewidth=2, label='Finitist (Linear)')

    ax1.axhline(y=1/np.sqrt(epsilon), color='green', linestyle=':', linewidth=2,
               label=f'Max Dilation = 1/√ε ≈ {1/np.sqrt(epsilon):.1f}')
    ax1.axvline(x=1.0, color='black', linestyle='--', linewidth=1.5, label='v = c')

    ax1.set_title("Time Dilation Factor γ vs. Velocity", fontsize=13, fontweight='bold')
    ax1.set_xlabel("β = v/c", fontsize=12)
    ax1.set_ylabel("Lorentz Factor γ", fontsize=12)
    ax1.set_ylim(0, 50)
    ax1.legend(fontsize=10)
    ax1.grid(True, alpha=0.3)

    # Right: Zoom near c
    beta_zoom = np.linspace(0.95, 1.0, 500)
    ax2.plot(beta_zoom, gamma_continuum(beta_zoom), 'r-', linewidth=1.5, alpha=0.5)
    ax2.plot(beta_zoom, gamma_finitist(beta_zoom, epsilon), 'g-', linewidth=2)
    ax2.axhline(y=1/np.sqrt(epsilon), color='green', linestyle=':', linewidth=1.5)
    ax2.axvline(x=1.0, color='black', linestyle='--', linewidth=1)

    ax2.set_title("Near Light Speed: Finite vs. Infinite Dilation", fontsize=12, fontweight='bold')
    ax2.set_xlabel("β = v/c", fontsize=11)
    ax2.set_ylabel("γ", fontsize=11)
    ax2.set_ylim(0, 10)
    ax2.grid(True, alpha=0.3)
    ax2.text(0.97, 8, f"Continuum: γ → ∞\nFinitist: γ_max = {1/np.sqrt(epsilon):.1f}",
            fontsize=10, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

    plt.suptitle("COMPUTATIONAL FINITISM: TIME DILATION IS FINITE AT v=c",
                fontsize=14, fontweight='bold', y=0.98)
    plt.tight_layout()

    fig.savefig("time_dilation_finitist.png", dpi=300, bbox_inches='tight')
    print("\n✓ Saved: time_dilation_finitist.png")
    plt.show()

    # Print physical implications
    print("\n" + "=" * 70)
    print("PHYSICAL IMPLICATIONS")
    print("=" * 70)
    print(f"• Maximum time dilation: γ_max = {1/np.sqrt(epsilon):.3f} (for d_max={d_max})")
    print(f"• For Planck-scale ε ~ 10^-35: γ_max ~ 10^17 (finite but enormous)")
    print("• Photons experience FINITE proper time, not zero time")
    print("• Penrose's 'end of time' argument relies on continuum infinity")
    print("• In finitism: time never stops, only dilates to maximum")
    print("=" * 70)

if __name__ == "__main__":
    plot_time_dilation()
```
